## 1.4 Type conversion (casting)

Sometimes data comes in the wrong type — e.g. a number typed as text from a spreadsheet or a form. You convert between types with `int()`, `float()`, `str()`, and `bool()`.

````python
age_text = "29"          # this looks like a number but it's a string
age_number = int(age_text)

print(age_text, type(age_text))
print(age_number, type(age_number))
print(age_number + 1)    # only works because we converted it first
````

## 1.5 Strings

Strings are how you'll work with names, categories, labels — anything textual in a dataset. They come with a rich set of built-in tools.

````python
product = "Wireless Headphones"

print(len(product))          # number of characters
print(product[0])            # first character (indexing starts at 0!)
print(product[-1])           # last character
print(product[0:8])          # slicing: characters from index 0 up to (not including) 8
print(product.upper())
print(product.lower())
print(product.replace("Wireless", "Bluetooth"))
````

**f-strings** are the modern, readable way to build strings that include variables. Put an `f` before the quotes and wrap variables in `{}`.

````python
customer = "Amara"
total = 14.25
items = 3

# The old, clunky way:
print("Receipt for " + customer + ": " + str(items) + " items, total $" + str(total))

# The f-string way -- read this out loud, it's much clearer:
print(f"Receipt for {customer}: {items} items, total ${total}")

# f-strings can also format numbers, e.g. 2 decimal places:
print(f"Receipt for {customer}: total ${total:.2f}")
````

A few more string methods you'll use constantly when cleaning real data:

````python
messy = "   San Francisco, CA   "

print(repr(messy.strip()))              # removes leading/trailing whitespace
print(messy.strip().split(","))         # splits into a list on a separator
print(", ".join(["red", "green", "blue"]))  # the reverse of split: joins a list into a string
````

## 1.7 Loops — repeating actions

A **`for` loop** repeats a block of code once per item in a sequence. This is the single most important tool for working with data — almost everything in `pandas`/`NumPy` is a faster, vectorized replacement for a `for` loop like the ones you're about to write.

````python
for number in range(5):     # range(5) -> 0, 1, 2, 3, 4
    print(number)
````

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(f"I like {fruit}")
````

A **`while` loop** repeats as long as a condition stays `True`. Use it when you don't know in advance how many times you'll repeat.

````python
count = 0
while count < 5:
    print(f"count is {count}")
    count = count + 1   # without this line, the loop would run forever!
````

`break` exits a loop early; `continue` skips to the next iteration.

````python
for n in range(10):
    if n == 3:
        continue   # skip printing 3
    if n == 7:
        break      # stop the loop entirely once we hit 7
    print(n)
````

## 2.1 Lists

A **list** is an ordered, changeable collection — the workhorse container in Python. Think of a single column of a spreadsheet.

````python
daily_sales = [1250.00, 980.50, 1430.25, 1100.00, 875.75]

print(daily_sales)
print(len(daily_sales))          # number of items
print(daily_sales[0])            # first item (index 0)
print(daily_sales[-1])           # last item
print(daily_sales[1:4])          # slice: items at index 1, 2, 3
````

Lists are **mutable** — you can change them after creating them.

````python
daily_sales.append(1600.00)      # add to the end
print(daily_sales)

daily_sales[0] = 1300.00         # overwrite an item
print(daily_sales)

daily_sales.sort()               # sort in place, ascending
print(daily_sales)

daily_sales.sort(reverse=True)
print(daily_sales)

removed = daily_sales.pop()      # removes & returns the last item
print("removed:", removed, "| remaining:", daily_sales)
````

Python has built-in functions that work directly on lists of numbers — these are exactly what NumPy will later do much faster on large datasets.

````python
print("sum:", sum(daily_sales))
print("min:", min(daily_sales))
print("max:", max(daily_sales))
print("average:", sum(daily_sales) / len(daily_sales))
````

## 2.3 Dictionaries

A **dictionary** stores `key: value` pairs. This is the closest pure-Python structure to a single *row* of a spreadsheet (each key is a column name), and it's exactly what `pandas` builds on.

````python
customer = {
    "name": "Amara Okafor",
    "age": 34,
    "member": True,
    "total_spent": 482.75,
}

print(customer["name"])              # access by key
print(customer.get("age"))           # .get() is a safer way to access -- won't crash if missing
print(customer.get("email", "N/A"))  # provide a default if the key doesn't exist

customer["email"] = "amara@example.com"   # add a new key
customer["total_spent"] += 25.00          # update a value
print(customer)
````

## 2.4 Sets

A **set** is an unordered collection of **unique** values — duplicates are automatically removed. Sets also directly implement the sample-space and event ideas from the probability class: `union` (OR), `intersection` (AND).

````python
purchased_categories = ["electronics", "books", "electronics", "home", "books", "toys"]

unique_categories = set(purchased_categories)
print(unique_categories)
print(len(unique_categories))

# Set operations map directly onto probability's AND / OR rules:
bought_online = {"Amara", "Chen", "Deja"}
bought_in_store = {"Chen", "Baraka"}

print("bought online OR in store (union):        ", bought_online | bought_in_store)
print("bought online AND in store (intersection):", bought_online & bought_in_store)
print("bought online but NOT in store (difference):", bought_online - bought_in_store)
````

## 2.5 List comprehensions

A **list comprehension** builds a new list from an existing one in a single, readable line. It's a compact `for` loop — and it's the mental stepping stone to NumPy's vectorized operations next week (where you won't need a loop *or* a comprehension — the whole array updates at once).

````python
prices = [10, 25, 8, 42, 15]

# The loop way:
prices_with_tax = []
for p in prices:
    prices_with_tax.append(p * 1.08)
print(prices_with_tax)

# The comprehension way -- same result, one line:
prices_with_tax_v2 = [p * 1.08 for p in prices]
print(prices_with_tax_v2)

# Comprehensions can filter too:
expensive = [p for p in prices if p > 15]
print(expensive)
````

## 2.6 Functions

A **function** packages up a piece of logic so you can reuse it without copy-pasting. You've been *calling* functions all along (`print()`, `len()`, `sum()`); now you'll *write* your own.

````python
def add_tax(price, tax_rate=0.08):     # tax_rate has a default value
    '''Return price with tax added.'''
    return price * (1 + tax_rate)

print(add_tax(100))            # uses the default 8% tax
print(add_tax(100, 0.05))      # overrides the default with 5%
````

- `def` starts a function definition; the code inside is indented.
- Values in `()` are **parameters**; a parameter can have a **default value** (`tax_rate=0.08`), making it optional when calling.
- `return` sends a value back to whoever called the function. Without `return`, a function gives back `None`.
- The text in triple quotes right after `def` is a **docstring** — documentation that tools (and `help()`) can read.

````python
help(add_tax)
````

Functions can return multiple values at once (as a tuple):

````python
def sales_summary(sales):
    '''Return (total, average, minimum, maximum) for a list of sales figures.'''
    total = sum(sales)
    average = total / len(sales)
    return total, average, min(sales), max(sales)

daily_sales = [1250.00, 980.50, 1430.25, 1100.00, 875.75]
total, avg, lo, hi = sales_summary(daily_sales)
print(f"Total: {total}, Average: {avg:.2f}, Min: {lo}, Max: {hi}")
````

## 3.2 Error handling

Real-world data is messy. Instead of letting one bad value crash your whole analysis, `try`/`except` lets you catch the error and decide what to do.

````python
# This will crash -- run it to see what a real error (a "traceback") looks like:
value = int("not a number")
````

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print(f"Warning: can't divide {a} by zero, returning None")
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

````python
def parse_price(text):
    try:
        return float(text)
    except ValueError:
        print(f"Warning: '{text}' isn't a valid price, skipping")
        return None

raw_prices = ["19.99", "24.50", "N/A", "8.00", "oops"]

clean_prices = []
for p in raw_prices:
    parsed = parse_price(p)
    if parsed is not None:
        clean_prices.append(parsed)

print(clean_prices)
print("total of valid prices:", sum(clean_prices))
````

````python
def safe_average(values):
    total = 0
    count = 0
    for v in values:
        try:
            total += v
            count += 1
        except TypeError:
            print(f"Warning: skipping non-numeric value {v!r}")

    if count == 0:
        print("Warning: no valid numbers to average")
        return None

    return total / count

print(safe_average([1, 2, 3, "oops", 4]))
print(safe_average([]))
````

## 3.4 Reading & writing CSV data

**CSV** (comma-separated values) is the most common plain-text format for tabular data, and it's what you'll load into `pandas` next week. Python's built-in `csv` module handles the fiddly bits (like commas inside quoted fields) for you.

````python
import csv

rows = [
    ["date", "product", "quantity", "price"],
    ["2026-01-05", "Wireless Mouse", 3, 19.99],
    ["2026-01-05", "USB-C Cable", 5, 8.50],
    ["2026-01-06", "Wireless Mouse", 1, 19.99],
    ["2026-01-06", "Keyboard", 2, 34.00],
    ["2026-01-07", "USB-C Cable", 10, 8.50],
]

with open("sales_data.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

print("sales_data.csv written.")
````

````python
# Reading it back with csv.reader gives you plain lists (including the header row)
with open("sales_data.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
````

````python
# csv.DictReader is usually nicer to work with: each row becomes a dictionary
# using the header row as keys -- this is conceptually a single step away from a
# pandas DataFrame.
with open("sales_data.csv", "r") as f:
    reader = csv.DictReader(f)
    sales_records = list(reader)

for record in sales_records[:3]:
    print(record)
````

## 3.5 Mini project: pure-Python data analysis

Let's put everything from these three classes together on a slightly bigger, messier dataset — the kind of thing you'll hand to `pandas` next week, but built and analyzed with nothing but the plain Python you now know.

````python
import random

random.seed(7)   # makes the "random" data reproducible for everyone running this notebook

products = ["Wireless Mouse", "USB-C Cable", "Keyboard", "Monitor Stand", "Webcam"]
prices = {"Wireless Mouse": 19.99, "USB-C Cable": 8.50, "Keyboard": 34.00,
          "Monitor Stand": 27.50, "Webcam": 45.00}

transactions = []
for i in range(200):
    product = random.choice(products)
    quantity = random.randint(1, 5)
    transactions.append({
        "id": i + 1,
        "product": product,
        "quantity": quantity,
        "price": prices[product],
        "revenue": round(quantity * prices[product], 2),
    })

print(f"Generated {len(transactions)} transactions.")
print(transactions[0])
````

````python
# Total revenue and average transaction size -- functions from Class 2
def total_revenue(txns):
    return sum(t["revenue"] for t in txns)

def average_transaction(txns):
    return total_revenue(txns) / len(txns)

print(f"Total revenue: ${total_revenue(transactions):,.2f}")
print(f"Average transaction: ${average_transaction(transactions):.2f}")
````

````python
# Revenue by product -- this is a "group by", done entirely with a dictionary and a loop.
# This exact operation is one line in pandas (.groupby()) -- worth remembering how much
# work it is by hand!
revenue_by_product = {}

for t in transactions:
    product = t["product"]
    revenue_by_product[product] = revenue_by_product.get(product, 0) + t["revenue"]

for product, revenue in sorted(revenue_by_product.items(), key=lambda x: x[1], reverse=True):
    print(f"{product:16s} ${revenue:,.2f}")
````

````python
# The best-selling product by total revenue
best_product = max(revenue_by_product, key=revenue_by_product.get)
print(f"Best-selling product by revenue: {best_product} (${revenue_by_product[best_product]:,.2f})")
````

````python
def mean(values):
    return sum(values) / len(values)

def variance(values):
    avg = mean(values)
    return sum((v - avg) ** 2 for v in values) / len(values)

quantities = [t["quantity"] for t in transactions]

print(f"Mean quantity: {mean(quantities):.2f}")
print(f"Std dev of quantity: {variance(quantities) ** 0.5:.2f}")
````

## 3.6 Why NumPy? (setting up next week)

Look back at everything that just took real effort in this mini project:

- `sum(t["revenue"] for t in txns)` — a manual loop to add things up
- `revenue_by_product.get(product, 0) + ...` — hand-rolled "group by"
- our own `mean()` / `variance()` — reimplementing basic statistics from scratch
- **type conversion everywhere** — CSV values are strings until you fix them, one field at a time

None of this is *wrong* — it's exactly what's happening under the hood in any data tool — but it's slow to write, easy to get subtly wrong (off-by-one loops, forgetting a conversion), and, for large datasets, **slow to run**: a plain Python `for` loop processes one item at a time.

**NumPy's whole point** is to fix this for numeric data:
- Store numbers in an `ndarray` (an "n-dimensional array") instead of a list.
- Operate on the *entire array at once* — no explicit loop — which is called **vectorization**.
- Run at C-level speed under the hood, often 10–100x faster than an equivalent Python loop on large data.

Here's a preview — don't worry about understanding every line, we build this from scratch next week:

````python
try:
    import numpy as np

    quantities_array = np.array(quantities)

    print("NumPy is installed. Version:", np.__version__)
    print("mean:  ", quantities_array.mean())     # compare to our hand-written mean()
    print("std:   ", quantities_array.std())
    print("sum:   ", quantities_array.sum())
    print("doubled (whole array, no loop!):", (quantities_array * 2)[:10])

except ImportError:
    print("NumPy isn't installed in this environment yet.")
    print("Before next week's class, install it with:  pip install numpy")
````

Notice: **no `for` loop**, no `.get()` bookkeeping, no manual `** 0.5`. That one-line `quantities_array.mean()` replaces the multi-line `mean()`/`variance()` functions you wrote by hand today. That's the whole pitch for NumPy, and it's why everything you practiced this week — variables, loops, conditionals, functions — is the foundation it's built on, not something you'll stop using.
